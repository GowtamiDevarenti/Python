```python
#Part 2: Go deeper

#Task 1: Central or not

central = ['Södermalms', 'Östermalms', 'Norrmalms', 'Kungsholmens']
mask = listings['neighbourhood_cleansed'].isin(central)
mask.sum()      # 2331
mask.mean()     # ~0.510

# share of private rooms inside / outside the centre
(listings[mask]['room_type'] == 'Private room').mean()    
(listings[~mask]['room_type'] == 'Private room').mean()    
```

```python
#Task 2: Room for a group
entire = listings['room_type'] == 'Entire home/apt'
big = listings['accommodates'] >= 4

big[entire].mean()      
big[~entire].mean()     
```

```python
#Task 3: Best reviewed, fairly
enough = listings[listings['number_of_reviews'] >= 50]
(enough['review_scores_rating'] == 5.0).sum()     # 14

cols = ['name', 'neighbourhood_cleansed', 'number_of_reviews', 'review_scores_rating']

top = enough.sort_values(['review_scores_rating', 'number_of_reviews'],
                         ascending=[False, False])[cols].head(5)
top

enough.nlargest(5, 'review_scores_rating')[cols]
```

```python
#Task 4: Label or position
top.loc[347, 'name']        # by label (index number)
top.iloc[3]['name']         # by position (4th row)
# both give "Stockholm"

top = top.reset_index(drop=True)
top.loc[347, 'name']        # KeyError
```
